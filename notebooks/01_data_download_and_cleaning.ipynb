## Resample to hourly

Energy readings are averaged over each hour (6 samples/hour; mean and sum differ only by a factor 6, so model rankings are unaffected). Hourly resolution makes SARIMA with daily seasonality (m=24) tractable.

```python
hourly = data.resample_hourly(raw)
print(hourly.shape)
hourly[[TARGET]].describe().round(1)
```
